```python
from zad1 import etykiety_wierzcholkow, generuj_siec, wczytaj_siec, wizualizuj_siec
from zad2 import ford_fulkerson, weryfikuj_przeplyw, wizualizuj_wynik, wypisz_historie

w, poj, et = wczytaj_siec("siec_1.txt")
fmax1, przep1, kroki1 = ford_fulkerson(len(et), poj, historia=True)
print("=== Sieć ===")
wypisz_historie(kroki1, et)
wizualizuj_wynik(w, poj, przep1, fmax1, et)

w_los, poj_los = generuj_siec(N=3, seed=None)
n_los = sum(len(g) for g in w_los)
et_los = etykiety_wierzcholkow(n_los)
wizualizuj_siec(w_los, poj_los, etykiety=et_los, tytul="Losowa sieć")
fmax_los, przep_los, kroki_los = ford_fulkerson(n_los, poj_los, historia=True)
assert weryfikuj_przeplyw(n_los, poj_los, przep_los)
print(f"\nLosowa sieć (N = 3): |f_max| = {fmax_los}")
wypisz_historie(kroki_los, et_los)
wizualizuj_wynik(w_los, poj_los, przep_los, fmax_los, et_los)

poj_przeciw = {(0, 1): 5, (0, 2): 5, (1, 2): 4, (2, 1): 3, (1, 3): 6, (2, 3): 4}
try:
    ford_fulkerson(4, poj_przeciw)
    assert False, "Łuki przeciwbieżne"
except ValueError:
    pass

fmax_p, przep_p = ford_fulkerson(4, poj_przeciw, rozbij_przeciwbiezne=True)
assert fmax_p == 10 and weryfikuj_przeplyw(4, poj_przeciw, przep_p, zezwalaj_przeciwbiezne=True)
```
